```python
import random

numsim = 1000000 #number of simulations
success = 0

for x in range(numsim):
    bag = ['R']*5 + ['B']*3
    draws = random.sample(bag, 2) # samples without replacement
    if 'R' in draws and 'B' in draws:
        success += 1

print(f"Monte Carlo Probability: {success / numsim:.4f} vs Analytical: {15/28:.4f}") # 15/28 calculated analytically
```
