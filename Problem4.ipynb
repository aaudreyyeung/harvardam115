```python
import random
from collections import Counter

numsim = 2000000 #number of simulations

sum_counts = Counter()
b_success = 0
c_success = 0
d_success = 0
countA1_B = Counter()
condition = 0
f_success = 0

for x in range(numsim):
    die1 = random.randint(1, 6)
    die2 = random.randint(1, 6)

    sum_counts[die1 + die2] += 1  # a) distribution of the sum

    if die1 == 1 or die2 == 1:  # b) at least one "one"
        b_success += 1

    if (die1, die2) in [(1, 6), (6, 1)]: # c) one "one", other "six"
        c_success += 1

    if die1 == 1 and die2 == 1:  # d) both dice "one"
        d_success += 1

    if die1 == 1:  # e) law of total probability
        countA1_B[die2] += 1

    if die1 == 6 or die2 == 6:  # f) given at least one "six"
        condition += 1
        if die1 == 1 or die2 == 1:
            f_success += 1

#results
analytical = {2:1/36, 3:2/36, 4:3/36, 5:4/36, 6:5/36, 7:6/36,
              8:5/36, 9:4/36, 10:3/36, 11:2/36, 12:1/36}
print("a) Sum distribution:")
for s in range(2, 13):
    print(f"  {s}: Monte Carlo = {sum_counts[s]/numsim:.4f} vs Analytical = {analytical[s]:.4f}")

print(f"\nb) Monte Carlo: {b_success/numsim:.4f} vs Analytical: {11/36:.4f}")
print(f"c) Monte Carlo: {c_success/numsim:.4f} vs Analytical: {2/36:.4f}")
print(f"d) Monte Carlo: {d_success/numsim:.4f} vs Analytical: {1/36:.4f}")

mc_sum = sum(countA1_B[b]/numsim for b in range(1, 7))
print(f"\ne) Sum of P(A=1,B=b): {mc_sum:.4f} vs Analytical: {1/6:.4f}")

print(f"\nf) Monte Carlo: {f_success/condition:.4f} vs Analytical: {2/11:.4f}")
```
