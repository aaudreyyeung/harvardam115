```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-3, 3, 500)
y = np.linspace(-3, 3, 500)
X, Y = np.meshgrid(x, y)
Z = X + 1j*Y

g_forward  = np.abs(1 + Z)
g_backward = np.abs(1 - Z)   #|1/(1-z)| <= 1  <=>  |1-z| >= 1

fig, ax = plt.subplots(1, 2, figsize=(10, 5))

ax[0].contourf(X, Y, g_forward <= 1, levels=[0.5, 1.5], colors=['blue'])
ax[0].set_title('Forward Euler stability region')

ax[1].contourf(X, Y, g_backward >= 1, levels=[0.5, 1.5], colors=['red'])
ax[1].set_title('Backward Euler stability region')

for a in ax:
    a.axhline(0, color='k', lw=0.5)
    a.axvline(0, color='k', lw=0.5)
    a.set_xlabel('Re(z)')
    a.set_ylabel('Im(z)')
    a.set_aspect('equal')

plt.tight_layout()
plt.show()
```
