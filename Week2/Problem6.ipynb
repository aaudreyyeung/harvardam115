```python
import numpy as np
p_values = np.linspace(0.001, 0.4, 400) # Define range for p
s_candidates = np.arange(1, 101) # Define candidate integer pool sizes from 1 to 100

results = []

for p in p_values:
    best_s = 1
    min_expected_per_capita = 1.0  # For s=1, expected tests per capita E(T)/N is exactly 1.0
    
    for s in s_candidates:
        if s == 1:
            continue
        expected_per_capita = (1 / s) + 1 - (1 - p)**s  # Calculate expected tests per capita: (1/s) + 1 - (1-p)^s
        
        if expected_per_capita < min_expected_per_capita:
            min_expected_per_capita = expected_per_capita
            best_s = s
            
    results.append((p, best_s, min_expected_per_capita))

print(f"{'Prevalence':<15}{'Optimal Pool Size':<25}{'Expected Tests per Capita'}")
for p, s, val in results[::40]: # Print every 40th step for brevity
    print(f"{p:<15.3f}{s:<25}{val:.4f}")
```
